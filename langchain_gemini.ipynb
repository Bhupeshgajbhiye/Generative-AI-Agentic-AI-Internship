Language Translation

```python
from langchain_core.messages import HumanMessage, SystemMessage

messages = [
    SystemMessage("Translate the following English to Marathi"),
    HumanMessage("how are you teams working on langchain with google gemini model")
]
model.invoke(messages)
```

```python
response1 = model.invoke(messages)
print(response1.text)
```

```python
from langchain_core.messages import HumanMessage, SystemMessage
messages = [
    SystemMessage("Translate the following Hindi to English"),
    HumanMessage("बहुत अच्छा टीम वर्क कर रहे हो आप लोग")
]

model.invoke(messages)
```

```python
response2 = model.invoke(messages)
print(response2.text)
```

```python
model.invoke("Hello")
model.invoke([{"role": "user", "content": "Hello"}])
model.invoke([HumanMessage("Hello")])
```

Prompt Templete

```python
from langchain_core.prompts import ChatPromptTemplate
system_template = " Translate the following from English into {language}"
prompt_template = ChatPromptTemplate.from_messages(
    [("system", system_template), ("user", "{text}")])
```

```python
prompt = prompt_template.invoke({"language": "Odia", "text": "how are you team"} )
prompt
```

```python
prompt.to_messages()
```

```python
response3 = model.invoke(prompt)
print(response3.content)
```

HUGGING FACE EMBADDING

```python
pip install -qU  langchain langchain-huggingface sentence_transformers
```

```python
from langchain_huggingface.embeddings import HuggingFaceEmbeddings
```

```python
embeddings = HuggingFaceEmbeddings(model_name="sentence-transformers/all-mpnet-base-v2") 
```

```python
text = "This is a text document." 
```

```python
query_result = embeddings.embed_query(text)
```

```python
print(query_result)
```
