Assignment 2:Print every token with its index.

```python
text = "I love learning NLP"
tokens = text.split()
for i, token in enumerate(tokens):
    print(i, token)
```

Assignment 3: Find the longest token in the sentence. Machine Learning is amazing.

```python
text = "Machine Learning is Amazing"
tokens = text.split()
longest = max(tokens, key=len)
print(longest)
```

Assignment 4: Print every token with its index.

```python
text = "Machine Learning is Amazing"
tokens = text.split()
shortest = min(tokens, key=len)
print(shortest)
```

Assignment 5 : Count frequency of every token.

```python
from collections import Counter
text = "AI AI AI NLP NLP"
tokens = text.split()
freq = Counter(tokens)
print(freq)
```

Assignment 6 : Find unique tokens.

```python
text = "AI AI NLP NLP Deep Learning"
tokens = text.split()
unique_tokens = set(tokens)
print(unique_tokens)
```

Assignment 7 : Question - Create vocabulary from text.

```python
text = "I love AI and NLP"
tokens = text.split()
vocab = sorted(set(tokens))
print(vocab)
```

Assignment 8 : Convert tokens into token IDs.

```python
text = "I love AI"
tokens = text.split()
vocab = {word:i for i, word in enumerate(set(tokens))}
token_ids = [vocab[word] for word in tokens]
print(token_ids)
```
